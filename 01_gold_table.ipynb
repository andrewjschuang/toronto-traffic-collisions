### Collisions segmented by date and time

- year (population number?)
- month (specific seasons?)
- dow (days of week?)
- time slot (rush hour, late at night?) - not possible, OCC_DATE is always 5 AM

```sql
CREATE OR REPLACE TABLE traffic_collisions_by_year_gold
AS
SELECT
    OCC_YEAR,
    COUNT_IF(FATALITIES > 0) AS FATALITIES,
    COUNT(CASE WHEN AUTOMOBILE IS TRUE THEN 1 END) AS AUTOMOBILE,
    COUNT(CASE WHEN MOTORCYCLE IS TRUE THEN 1 END) AS MOTORCYCLE,
    COUNT(CASE WHEN PASSENGER IS TRUE THEN 1 END) AS PASSENGER,
    COUNT(CASE WHEN BICYCLE IS TRUE THEN 1 END) AS BICYCLE,
    COUNT(CASE WHEN PEDESTRIAN IS TRUE THEN 1 END) AS PEDESTRIAN,
    COUNT(CASE WHEN INJURY_COLLISIONS IS TRUE THEN 1 END) AS INJURY_COLLISIONS,
    COUNT(CASE WHEN FTR_COLLISIONS IS TRUE THEN 1 END) AS FTR_COLLISIONS,
    COUNT(CASE WHEN PD_COLLISIONS IS TRUE THEN 1 END) AS PD_COLLISIONS
FROM workspace.default.traffic_collisions_silver
GROUP BY OCC_YEAR
ORDER BY OCC_YEAR ASC
;
```

```sql
CREATE OR REPLACE TABLE traffic_collisions_by_year_month_gold
AS
SELECT
    OCC_YEAR,
    OCC_MONTH,
    COUNT_IF(FATALITIES > 0) AS FATALITIES,
    COUNT(CASE WHEN AUTOMOBILE IS TRUE THEN 1 END) AS AUTOMOBILE,
    COUNT(CASE WHEN MOTORCYCLE IS TRUE THEN 1 END) AS MOTORCYCLE,
    COUNT(CASE WHEN PASSENGER IS TRUE THEN 1 END) AS PASSENGER,
    COUNT(CASE WHEN BICYCLE IS TRUE THEN 1 END) AS BICYCLE,
    COUNT(CASE WHEN PEDESTRIAN IS TRUE THEN 1 END) AS PEDESTRIAN,
    COUNT(CASE WHEN INJURY_COLLISIONS IS TRUE THEN 1 END) AS INJURY_COLLISIONS,
    COUNT(CASE WHEN FTR_COLLISIONS IS TRUE THEN 1 END) AS FTR_COLLISIONS,
    COUNT(CASE WHEN PD_COLLISIONS IS TRUE THEN 1 END) AS PD_COLLISIONS
FROM workspace.default.traffic_collisions_silver
GROUP BY
    OCC_YEAR,
    OCC_MONTH
ORDER BY
    OCC_YEAR ASC,
    TO_DATE(OCC_MONTH, 'MMMM') ASC
;
```

```sql
CREATE OR REPLACE TABLE traffic_collisions_by_year_month_dow_gold
AS
SELECT
    OCC_YEAR,
    OCC_MONTH,
    OCC_DOW,
    COUNT_IF(FATALITIES > 0) AS FATALITIES,
    COUNT(CASE WHEN AUTOMOBILE IS TRUE THEN 1 END) AS AUTOMOBILE,
    COUNT(CASE WHEN MOTORCYCLE IS TRUE THEN 1 END) AS MOTORCYCLE,
    COUNT(CASE WHEN PASSENGER IS TRUE THEN 1 END) AS PASSENGER,
    COUNT(CASE WHEN BICYCLE IS TRUE THEN 1 END) AS BICYCLE,
    COUNT(CASE WHEN PEDESTRIAN IS TRUE THEN 1 END) AS PEDESTRIAN,
    COUNT(CASE WHEN INJURY_COLLISIONS IS TRUE THEN 1 END) AS INJURY_COLLISIONS,
    COUNT(CASE WHEN FTR_COLLISIONS IS TRUE THEN 1 END) AS FTR_COLLISIONS,
    COUNT(CASE WHEN PD_COLLISIONS IS TRUE THEN 1 END) AS PD_COLLISIONS
FROM workspace.default.traffic_collisions_silver
GROUP BY
    OCC_YEAR,
    OCC_MONTH,
    OCC_DOW
ORDER BY
    OCC_YEAR ASC,
    TO_DATE(OCC_MONTH, 'MMMM') ASC,
    CASE UPPER(OCC_DOW)
        WHEN 'SUNDAY' THEN 1
        WHEN 'MONDAY' THEN 2
        WHEN 'TUESDAY' THEN 3
        WHEN 'WEDNESDAY' THEN 4
        WHEN 'THURSDAY' THEN 5
        WHEN 'FRIDAY' THEN 6
        WHEN 'SATURDAY' THEN 7
    END ASC
;
```
