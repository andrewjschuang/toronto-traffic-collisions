### Collisions by location

Collisions segmented by neighborhood

Note: would be interesting to join with a dataset with population count for each neighbourhood

```sql
CREATE OR REPLACE TABLE workspace.default.traffic_collisions_by_neighbourhood_gold
AS
SELECT
    NEIGHBOURHOOD_158,
    COUNT_IF(FATALITIES > 0) AS FATALITIES,
    COUNT(CASE WHEN AUTOMOBILE IS TRUE THEN 1 END) AS AUTOMOBILE,
    COUNT(CASE WHEN MOTORCYCLE IS TRUE THEN 1 END) AS MOTORCYCLE,
    COUNT(CASE WHEN PASSENGER IS TRUE THEN 1 END) AS PASSENGER,
    COUNT(CASE WHEN BICYCLE IS TRUE THEN 1 END) AS BICYCLE,
    COUNT(CASE WHEN PEDESTRIAN IS TRUE THEN 1 END) AS PEDESTRIAN,
    COUNT(CASE WHEN INJURY_COLLISIONS IS TRUE THEN 1 END) AS INJURY_COLLISIONS,
    COUNT(CASE WHEN FTR_COLLISIONS IS TRUE THEN 1 END) AS FTR_COLLISIONS,
    COUNT(CASE WHEN PD_COLLISIONS IS TRUE THEN 1 END) AS PD_COLLISIONS
FROM workspace.default.traffic_collisions_silver
GROUP BY 1
ORDER BY CAST(regexp_extract(NEIGHBOURHOOD_158, r'\((\d+)\)', 1) AS INT) ASC
;
```
